```python
import numpy as np

# TASK 1
np.random.seed(42)
students = 10

names = np.random.choice(
    ["rakesh", "kumar", "jujjuri", "srinu", "dinakar"],
    size=students
)
age = np.random.randint(15, 18, size=students)
maths = np.random.randint(40, 100, size=students)
science = np.random.randint(40, 100, size=students)
english = np.random.randint(40, 100, size=students)

# FIXED shape
marks = np.array([maths, science, english]).T

print(marks.dtype)
print(marks.shape)

# TASK 2
print("\nOriginal Marks:\n", marks)

reshaped_marks = marks.reshape(2, 5, 3)
print("\nReshaped Marks:\n", reshaped_marks)

print("\nFirst 5 students (Maths & Science):\n", marks[:5, :2])

# (2, 5, 3) means:
# 2 groups, 5 students per group, 3 subjects per student

# TASK 3
with open("student_data.txt", "w") as file:
    file.write("Name | Age | Maths | Science | English\n")

    for i in range(students):
        file.write(
            f"{names[i]} | {age[i]} | {maths[i]} | {science[i]} | {english[i]}\n"
        )

print("\n✅ All student records saved successfully!")

# READ FILE
with open("student_data.txt", "r") as file:
    print("\n📄 File Contents:\n")
    for line in file:
        print(line.strip())
```
